```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine

url = "https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv"

# PostgreSQL connection
engine = create_engine(
    'postgresql+psycopg://root:root@localhost:5432/ny_taxi'
)

# Extract
df = pd.read_csv(url)

# Preview
print(df.head())
```

```python
# Check the SQL schema
print(
    pd.io.sql.get_schema(
        df,
        name='zones',
        con=engine
    )
)
```

```python
# Load
from tqdm.auto import tqdm
df.to_sql(
    name='zones',
    con=engine,
    if_exists='replace',
    index=False
)

print("Zones data successfully loaded into PostgreSQL.")
```
